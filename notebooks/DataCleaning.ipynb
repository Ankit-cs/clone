```python
df['agent'] = df['agent'].fillna(0)
```

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
```

```python
df['children'] = df['children'].fillna(0)
```

```python
df.isnull().sum()
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'])
```

```python
df.info()
```

```python
df['total_guests'] = (
    df['adults']
    + df['children']
    + df['babies']
)
```

```python
df['total_nights'] = (
    df['stays_in_week_nights']
    + df['stays_in_weekend_nights']
)
```

```python
import numpy as np
df['family'] = np.where(
    df['children'] + df['babies'] > 0,
    1,
    0
)
```

```python
months = {
    'January':1,
    'February':2,
    'March':3,
    'April':4,
    'May':5,
    'June':6,
    'July':7,
    'August':8,
    'September':9,
    'October':10,
    'November':11,
    'December':12
}

df['arrival_month_num'] = df['arrival_date_month'].map(months)
```

```python
def season(month):

    if month in [12,1,2]:
        return 'Winter'

    elif month in [3,4,5]:
        return 'Spring'

    elif month in [6,7,8]:
        return 'Summer'

    else:
        return 'Autumn'


df['season'] = df['arrival_month_num'].apply(season)
```
